```python
import pandas as pd
import numpy as np
```

```python
# read parquet file
df = pd.read_parquet('data/brasil_dengue_clima_final.parquet')
df
```

```python
# get the number of records per 'NOMBRE_MUNICIPIO','NU_ANO','MES' as dictionary
records_per_municipality = df.groupby(['NOMBRE_MUNICIPIO', 'NU_ANO', 'MES']).size().to_dict()
records_per_municipality
```

```python

#* Agrupar por municipio, año y mes, con la clasificacion adecuada
#* se debe realizar un comparativo que permita identificar municipios similares, a partir de condiciones climaticas
#* tambien se debe tener en cuenta el ambiente de la zona, urbno, rural, costero o del interior
```

```python
# actualizacion de codigos
df['CLASIFICACION'] = df['CLASIFICACION'].replace({1: 10, 2: 11, 3: 12, 4: 12})

# conteo de clasificaciones por año en diccionario
classification_counts = df.groupby(['NU_ANO', 'CLASIFICACION']).size().to_dict()
classification_counts
```

```python
df['CLASIFICACION'].value_counts()
```

```python
# filtrar por clasificacion 10, 11 y 12
df = df[df['CLASIFICACION'].isin([10, 11, 12])]
df
```

```python
# agrupar por municipio, año y mes y contar el numero de registros por cada grupo
grouped = df.groupby(['NOMBRE_MUNICIPIO', 'NU_ANO', 'MES']).size().reset_index(name='counts')
grouped
```
